```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
```

```python
tickers = ['aapl','amzn','nflx','fb','goog']
a = []
for ticker in tickers:
  df = pd.read_csv(f'{ticker}.csv')
  df['ticker'] = ticker.upper()
  a.append(df)
```

```python
new = pd.concat(a, ignore_index=True)
print(new.head())
print(new.tail())
```

```python
new.to_csv('faanf.csv',index=False)
```

```python
df = pd.read_csv('faanf.csv')
df['date'] = pd.to_datetime(df['date'])
df['volume'] = df['volume'].astype(int)
df = df.sort_values(['date','ticker']).reset_index(drop=True)

print(df.dtypes)
print(df.head)
```

```python
highest = df.nlargest(7,'volume')
print(highest)
```

```python
new_melted = df.melt(
    id_vars = ['date','ticker'],
    var_name = 'variable',
    value_name = 'value'
)
print(new_melted.head(5))
```
