```python
import numpy as np
import pandas as pd
PREVIOUS_MAX_ROWS = pd.options.display.max_rows
pd.options.display.max_rows = 20
np.random.seed(12345)
import matplotlib.pyplot as plt
plt.rc('figure', figsize=(10, 6))
np.set_printoptions(precision=4, suppress=True)
```

```python
df = pd.read_csv('parsed.csv')
```

a. Find the 95th percentile of earthquake magnitude in Japan using magType of 'mb'

```python
mb_Japan = df[(df['parsed_place']=='Japan') & (df['magType'] == 'mb')]
a = mb_Japan ['mag'].quantile(0.95)
print(a)
```

b. Find the percentage of earthquakes in Indonesia that were coupled with tsunamis.

```python
tsu_Indonesia = df[(df['parsed_place']=='Indonesia') & (df['tsunami']== 1)]
b = len(tsu_Indonesia) / len(df[(df['parsed_place']=='Indonesia')]) * 100
print(b)
```

c. Get summary statistics for earthquakes in Nevada

```python
e_Nevada = df[df['parsed_place']=='Nevada']
print(e_Nevada.describe())
```

d. Add a column to the dataframe indicating whether or not the earthquake happened in a country or US state that is on the Ring of Fire. Use Bolivia, Chile, Ecuador, Peru, Costa Rica, Guatemala, Mexico (be sure not to select New Mexico), Japan, Philippines, Indonesia, New Zealand, Antarctica (look for the Antarctic), Canada, Fiji, Alaska, Washington, California, Russia, Taiwan, Tonga, and the Kermadec Islands).

```python
ring_of_fire = ['Bolivia','Chile','Ecuador', 'Peru', 'Costa Rica', 'Guatemala', 'Mexico',
                'Japan', 'Philippines', 'Indonesia', 'New Zealand', 'Antarctic' , 'Canada',
                'Fiji', 'Alaska', 'Washington', 'California', 'Russia', 'Taiwan', 'Tonga', 'the Kermadec Islands']
df['ring_of_fire'] = df['parsed_place'].isin(ring_of_fire)
print(df['ring_of_fire'])
```

e. Calculate the number of earthquakes in the Ring of Fire locations and the number outside of them

```python
ear_rof = df[df['ring_of_fire'] == True]
print("the number of earthquakes in the Ring of Fire locations:", len(ear_rof))
ear_outrof = df[df['ring_of_fire'] == False]
print("the number of earthquakes outside of the Ring of Fire locations:", len(ear_outrof))
```

f. Find the tsunami count along the Ring of Fire.

```python
tsu_rof = df[(df['ring_of_fire'] == True) & (df['tsunami'] == 1)]
print(len(tsu_rof))
```
