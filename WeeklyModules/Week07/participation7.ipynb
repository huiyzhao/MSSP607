```python
import numpy as np
import pandas as pd
PREVIOUS_MAX_ROWS = pd.options.display.max_rows
pd.options.display.max_rows = 20
np.random.seed(12345)
import matplotlib.pyplot as plt
import matplotlib
plt.rc('figure', figsize=(10, 6))
np.set_printoptions(precision=4, suppress=True)
```

```python
df = pd.read_csv('InternetSales.csv', encoding="latin1", low_memory=False)
```

```python
print(df.columns)
```

```python
print(df.head())
```

```python
df['OrderDate']=pd.to_datetime(df['OrderDate'], format='%m/%d/%Y')
df['Year']=df['OrderDate'].dt.year
yearly=df.groupby('Year')['SalesAmount'].sum().reset_index()
```

```python
yearly
```

```python
fig=plt.figure(figsize=(10, 6))
ax=fig.add_subplot(1,1,1)
ax.bar(yearly["Year"].astype(str), yearly["SalesAmount"],
       color="k", alpha=0.7)
ax.set_title("Yearly Sales")
ax.set_xlabel("Year")
ax.set_ylabel("Total Sales")

fig.savefig("Yearly_Sales.pdf", bbox_inches="tight")
```
