
In the context of downloading data from Yahoo Finance using the yfinance library, the period parameter defines the range of data to download and can have the following values:

*    1d, 5d: data for the past 1 or 5 days

*   1mo, 3mo, 6mo: data for the past 1, 3, or 6 months
*   1y, 2y, 5y, 10y, ytd: data for the past 1, 2, 5, or 10 years, or year to date.


*   max: data for the maximum available period.


These values allow you to specify the time frame for which you want to retrieve historical data for a given stock.

#Task 1: Read into seperate data frame the following tickers: Apple, Facebook, Google, and Netflix; use Yahoo finance.

```python
import os
import subprocess

owner = "huiyzhao"
repository = "MSSP607"

repo_url = f"https://github.com/huiyzhao/MSSP607.git"

if not os.path.exists (repository):
  subprocess.run(["git", "clone", repo_url], check=True)

os.chdir(repository)
print("Current folder:", os.getcwd())
```

```python
apple = yf.Ticker("AAPL")
facebook = yf.Ticker("META")
google = yf.Ticker("GOOG")
netflix = yf.Ticker("NFLX")

apple_df = apple.history(period="max")
facebook_df = facebook.history(period="max")
google_df = google.history(period="max")
netflix_df = netflix.history(period="max")
```

#Task 2:  Join the individual stock data frames into a single data frame.

```python
all = pd.concat([
    apple_df.assign(Ticker="AAPL"),
    facebook_df.assign(Ticker="META"),
    google_df.assign(Ticker="GOOG"),
    netflix_df.assign(Ticker="NFLX")
]).reset_index()
all.head()
```

# Task 3: Create a visualization of the following financial metrics for each stock:  quickRatio, currentRatio, totalRevenue, debtToEquity, revenuePerShare, returnOnAssets, returnOnEquity, grossProfits


```python
metrics = ["quickRatio",
           "currentRatio",
           "totalRevenue",
           "debtToEquity",
           "revenuePerShare",
           "returnOnAssets",
           "returnOnEquity",
           "grossProfits"]
metrics = pd.DataFrame({
    "AAPL":[apple.info.get(metric) for metric in metrics],
    "META":[facebook.info.get(metric) for metric in metrics],
    "GOOG":[google.info.get(metric) for metric in metrics],
    "NFLX":[netflix.info.get(metric) for metric in metrics],
}, index=metrics)
metrics
```

```python
metrics2 = metrics.apply(pd.to_numeric, errors="coerce")
scaled = metrics2.apply(
    lambda row: (row-row.min()) / (row.max()-row.min()), axis=1
)
plt.figure(figsize=(10,7))
sns.heatmap(
    scaled,
    annot=True,
    fmt=".2f",
    cmap="YlGnBu",
    linewidth=0.5
)
plt.xlabel("Ticker")
plt.ylabel("Metric")
```

#Task 4: Create a single boxplot for the Closing price for each stock.


```python
plt.figure(figsize=(10,6))
sns.boxplot(data=all, x="Ticker",y="Close",color="#A7C7C5")
plt.xlabel("Ticker")
plt.ylabel("Closing Price")
```

# Task 5: Create a daily trading volume visualization for the year 2023 for each stock.


```python
all["Date"] = pd.to_datetime(all["Date"])
volume_2023 = all[all["Date"].dt.year == 2023]
plt.figure(figsize=(12,6))
sns.lineplot(data=volume_2023, x="Date",y="Volume",hue="Ticker",linewidth=0.5)
plt.xlabel("Date")
plt.ylabel("Trading Volume")
```
