```python
df = pd.read_csv('data_academic_performance.csv')
```

```python
print(df.columns.tolist())
print(df['GENDER'].value_counts(dropna=False))
print(df.isnull().sum())
```

```python
df = df.drop(columns = ['Unnamed: 9'])
```

```python
scores_pro = ['QR_PRO', 'CR_PRO', 'CC_PRO', 'ENG_PRO', 'WC_PRO', 'FEP_PRO', 'G_SC']
print(df[scores_pro].describe().round(2))
```

```python
print(df.groupby('GENDER')[scores_pro].mean().round(2))
```

```python
print(df.groupby('STRATUM')[scores_pro].mean().round(2))
```

```python
print(df.groupby('EDU_FATHER')[scores_pro].mean().round(2))
```

```python
print(df.groupby('EDU_MOTHER')[scores_pro].mean().round(2))
```

```python
print(df.groupby('SISBEN')[scores_pro].mean().round(2))
```

```python
from scipy import stats
male = df[df['GENDER'] == 'M']['G_SC']
female = df[df['GENDER'] == 'F']['G_SC']
t,p = stats.ttest_ind(male, female, equal_var = False)
print("Male mean:", round(male.mean(),2))
print("Female mean:", round(female.mean(),2))
print("Male SD:", round(male.std(),2))
print("Female SD:", round(female.std(),2))
print("Difference:", round(male.mean()-female.mean(),2))
print("t-statistic:",round(t,3))
print("p-value:", p)
```

```python
s1 = df[df['STRATUM']=='Stratum 1']['G_SC']
s2 = df[df['STRATUM']=='Stratum 2']['G_SC']
s3 = df[df['STRATUM']=='Stratum 3']['G_SC']
s4 = df[df['STRATUM']=='Stratum 4']['G_SC']
s5 = df[df['STRATUM']=='Stratum 5']['G_SC']
s6 = df[df['STRATUM']=='Stratum 6']['G_SC']

print("S1:", len(s1), s1.mean())
print("S2:", len(s2), s2.mean())
print("S3:", len(s3), s3.mean())
print("S4:", len(s4), s4.mean())
print("S5:", len(s5), s5.mean())
print("S6:", len(s6), s6.mean())

f,p = stats.f_oneway(s1,s2,s3,s4,s5,s6)
print("f-statistic:",round(f,3))
print("p-value:",p)
```

```python
for score in scores_pro:
  s1 = df[df['STRATUM'] == 'Stratum 1'][score]
  s6 = df[df['STRATUM'] == 'Stratum 6'][score]
  print(f"{score}: Stratum 1 = {s1.mean():.2f}, Stratum 6 = {s6.mean():.2f},gap = {s6.mean()-s1.mean():.2f}")
```

```python
ef1 = df[df['EDU_FATHER']=='Ninguno']['G_SC']
ef2 = df[df['EDU_FATHER']=='Complete primary ']['G_SC']
ef3 = df[df['EDU_FATHER']=='Complete Secundary']['G_SC']
ef4 = df[df['EDU_FATHER']=='Complete technique or technology']['G_SC']
ef5 = df[df['EDU_FATHER']=='Complete professional education']['G_SC']
ef6 = df[df['EDU_FATHER']=='Postgraduate education']['G_SC']

print("Ninguno:",len(ef1),ef1.mean())
print("Primary:",len(ef2),ef2.mean())
print("Secondary:",len(ef3),ef3.mean())
print("Technique or technology:",len(ef4),ef4.mean())
print("Professional education:",len(ef5),ef5.mean())
print("Postgraduate:",len(ef6),ef6.mean())

f,p = stats.f_oneway(ef1,ef2,ef3,ef4,ef5,ef6)
print("f-statistic:",round(f,3))
print("p-value:",p)
```
