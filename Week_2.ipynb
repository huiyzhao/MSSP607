```python
spam = [2,4,6,8,10]
spam[2]='hello'
print(spam)
```

```python
letters=["a","b","c","d"]

result=letters[int(int('3'*2)//11)]
print(result)

result_c=letters[-1]
print(result_c)

result_d=letters[:2]
print (result_d)
```

'3'*2 means to repeat the string '3', so we can get '33'
int('33') means to convert the string '33' to the integer 33
33//11 = 3
letters[3] = d

negative indexing starts from the end of the list, so -1 refers to the last element

[:2] starts from index 0 up to index 2, excluding index 2
